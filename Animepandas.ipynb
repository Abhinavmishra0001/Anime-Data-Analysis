```python
import pandas as pd
import numpy as np
df=pd.read_csv('/Users/abhinavmishra/Downloads/YTMLPROJECT/anime.csv')
```

```python
#extract episodes in a new column
def checkepisodes(txt):
    check = False
    data = ""
    for i in txt:
        if i == ")":
            check = False
            return data
        if check == True:
            data=data+i
        if i == '(':
            check = True
```

```python
df["Episodes"]= df["Title"].apply(checkepisodes)
df
```

```python
df["Episodes"]= df["Episodes"].str.replace(" eps","")
df["Episodes"]=df["Episodes"].astype(int)
```

```python
#extract dates in new column
def extractdates(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i]== ")":
           for j in range (i+1,i+20):
            data+=txt[j]
               
    return data
```

```python
df["Total_Time"] = df["Title"].apply(extractdates)
df["Start_Month"] = df["Total_Time"].str[:3]
df["End_Month"]=df["Total_Time"].str[10:14]
df["Start_Year"] = df["Total_Time"].str[4:9]
df["End_Year"] = df["Total_Time"].str[14:19]
```
